# Model B — `deduction_pct`

Reproduces the training path for **Model B** of Solvent's forecaster stage (Razorpay AI Buildathon, Track 04 — AI Finance Controller) for exploration and documentation. This is an **educational reproduction** of `backend/forecaster/train.py` + `backend/forecaster/features.py` — the canonical production model that the API actually serves is trained by running `python -m backend.forecaster.train` from the repo root, which persists its artifact to `backend/forecaster/models/model_b_deduction_pct.joblib` and its metrics to Postgres (`forecaster_metrics` table). This notebook uses the same algorithm, the same feature set, and the same `random_state`/`test_size`, reading from the flat CSV export in `../data/base_transactions.csv` instead of the live database, so it can be run standalone.

**What it predicts:** `deduction_pct` — the fraction of `txn_amount` deducted before settlement (gateway fee + GST-on-fee + any refund), as a 0-1 value.

**Model:** plain `sklearn.linear_model.LinearRegression` (closed-form OLS, no gradient descent, no LLM), inside the same kind of `ColumnTransformer` pipeline as Model A, with Model B's own feature set.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.compose import ColumnTransformer
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import FunctionTransformer, OneHotEncoder, PolynomialFeatures
from sklearn.model_selection import train_test_split, learning_curve
import joblib
from pathlib import Path

RANDOM_STATE = 42
TEST_SIZE = 0.2
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3
```

## 1. Load the training data

The same 257-row `base_transactions` table Model A trains on.

**Note on row order:** the CSV preserves the exact row order returned by `db.load_base_transactions()` (Postgres's default read order, not sorted by `txn_id`). `train_test_split`'s row-to-split assignment depends on input row order for a given `random_state`, so this order is what makes the metrics below reproduce `docs/metrics.md` exactly rather than just approximately.

```python
df = pd.read_csv('../data/base_transactions.csv')
bool_cols = ['is_weekend_or_holiday', 'had_dispute_flag', 'had_refund', 'gst_on_fee_flag']
df[bool_cols] = df[bool_cols].astype(int)
print(df.shape)
df.head()
```

## 2. Feature set

Matches `MODEL_B_NUMERIC` / `MODEL_B_CATEGORICAL` / `MODEL_B_TARGET` in `backend/forecaster/features.py` exactly.

```python
NUMERIC = ['txn_amount', 'had_refund', 'refund_amount', 'gst_on_fee_flag']
CATEGORICAL = ['payment_method']
TARGET = 'deduction_pct'

X = df[NUMERIC + CATEGORICAL]
y = df[TARGET]
print(y.describe())
```

## 3. Quick EDA — the target is effectively bimodal

Before fitting anything: how does `deduction_pct` relate to `had_refund`? This turns out to be the single most important thing about this dataset.

```python
corr = df['deduction_pct'].corr(df['had_refund'])
print(f'corr(deduction_pct, had_refund) = {corr:.4f}')
df.groupby('had_refund')['deduction_pct'].describe()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4))
axes[0].hist(y, bins=25, color='#5b8def', edgecolor='white')
axes[0].set_xlabel('deduction_pct')
axes[0].set_title('Target distribution (note the two clusters)')

df.boxplot(column='deduction_pct', by='had_refund', ax=axes[1])
axes[1].set_xlabel('had_refund')
axes[1].set_title('deduction_pct by had_refund')
plt.suptitle('')
fig.tight_layout()
plt.show()
```

Non-refunded transactions cluster tightly near 0.01-0.03 (ordinary gateway-fee deduction); refunded transactions jump to 0.68-0.99 (the refund consumes most of the settlement). A single linear model has to fit both regimes with one coefficient set — keep this in mind when reading the MAPE below.

## 4. Train / test split

Same `random_state=42`, `test_size=0.2` as `backend/forecaster/train.py` and the Model A notebook, so results are directly comparable to `docs/metrics.md`.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=TEST_SIZE, random_state=RANDOM_STATE
)
print(f'train: {len(X_train)}  test: {len(X_test)}')
```

## 5. Pipeline: `ColumnTransformer` + `LinearRegression`

```python
def build_pipeline(numeric_features, categorical_features, amount_poly_degree=1):
    other_numeric = [f for f in numeric_features if f != 'txn_amount']
    amount_transformer = Pipeline([
        ('poly', PolynomialFeatures(degree=amount_poly_degree, include_bias=False))
    ])
    transformers = [('amount', amount_transformer, ['txn_amount'])]
    if other_numeric:
        transformers.append(
            ('numeric', FunctionTransformer(feature_names_out='one-to-one'), other_numeric)
        )
    if categorical_features:
        transformers.append(
            ('categorical', OneHotEncoder(handle_unknown='ignore'), categorical_features)
        )
    preprocessor = ColumnTransformer(transformers)
    return Pipeline([('preprocess', preprocessor), ('model', LinearRegression())])


def mae(y_true, y_pred):
    return float(np.mean(np.abs(np.asarray(y_true) - np.asarray(y_pred))))


def mape(y_true, y_pred):
    y_true = np.asarray(y_true, dtype=float)
    y_pred = np.asarray(y_pred, dtype=float)
    nonzero = np.abs(y_true) > 1e-6
    if not nonzero.any():
        return float('nan')
    return float(np.mean(np.abs((y_true[nonzero] - y_pred[nonzero]) / y_true[nonzero])) * 100)
```

```python
baseline = build_pipeline(NUMERIC, CATEGORICAL, amount_poly_degree=1)
baseline.fit(X_train, y_train)
baseline_pred = baseline.predict(X_test)

baseline_mae = mae(y_test, baseline_pred)
baseline_mape = mape(y_test, baseline_pred)
print(f'Baseline linear  ->  MAE: {baseline_mae:.4f}   MAPE: {baseline_mape:.2f}%')
```

## 6. Curvature check

Same diagnostic as Model A: correlate residuals with `txn_amount²`. As the EDA above suggests, the real structure here is bimodality driven by `had_refund`, not curvature in `txn_amount` — so this check is expected to come back below threshold.

```python
residuals_train = y_train.values - baseline.predict(X_train)
amount_sq = X_train['txn_amount'].values ** 2
curvature_corr = np.corrcoef(residuals_train, amount_sq)[0, 1]
curvature_detected = bool(abs(curvature_corr) > 0.15)
print(f'corr(residuals, txn_amount^2) = {curvature_corr:.4f}')
print(f'curvature_detected = {curvature_detected}')
```

```python
chosen_pipeline, chosen_mae, chosen_mape, chosen_label = baseline, baseline_mae, baseline_mape, 'linear'

if curvature_detected:
    poly = build_pipeline(NUMERIC, CATEGORICAL, amount_poly_degree=2)
    poly.fit(X_train, y_train)
    poly_pred = poly.predict(X_test)
    poly_mae, poly_mape = mae(y_test, poly_pred), mape(y_test, poly_pred)
    print(f'Polynomial deg=2 ->  MAE: {poly_mae:.4f}   MAPE: {poly_mape:.2f}%')
    if poly_mae < baseline_mae:
        chosen_pipeline, chosen_mae, chosen_mape, chosen_label = poly, poly_mae, poly_mape, 'polynomial(degree=2)'
else:
    print('No curvature above threshold — polynomial variant not attempted (matches production run).')
    print('A degree-2 term on txn_amount would not fix this model anyway: the real')
    print('structure is the had_refund bimodality shown in the EDA above, not amount curvature.')

print(f'\nCHOSEN: {chosen_label}  ->  MAE: {chosen_mae:.4f}   MAPE: {chosen_mape:.2f}%')
```

## 7. Why there's no loss-vs-epoch curve

Same reason as Model A: closed-form OLS, one step, no epochs. The learning curve below is the honest substitute — held-out error vs. training-set size.

```python
train_sizes, train_scores, val_scores = learning_curve(
    build_pipeline(NUMERIC, CATEGORICAL, amount_poly_degree=1),
    X_train, y_train,
    train_sizes=np.linspace(0.2, 1.0, 8),
    cv=5,
    scoring='neg_mean_absolute_error',
    random_state=RANDOM_STATE,
)
train_mae_curve = -train_scores.mean(axis=1)
val_mae_curve = -val_scores.mean(axis=1)

plt.figure(figsize=(6, 4))
plt.plot(train_sizes, train_mae_curve, 'o-', color='#5b8def', label='Train MAE')
plt.plot(train_sizes, val_mae_curve, 'o-', color='#2bd6a3', label='Cross-val MAE')
plt.xlabel('Training examples')
plt.ylabel('MAE (deduction_pct)')
plt.title('Learning curve — no epochs, closed-form OLS')
plt.legend(frameon=False)
plt.tight_layout()
plt.show()
```

## 8. Predicted vs. actual, and residual diagnostics (held-out test set)

Watch the predicted-vs-actual plot in particular: predictions should cluster near ~0.0 and ~0.65-0.68 rather than tracking the true spread within the refund group (which runs up to ~0.99) — the model has learned `had_refund` is the dominant signal, but one linear coefficient on it can't also capture the variance *within* the refunded group. This is the direct explanation for the high MAPE despite low MAE.

```python
y_pred = chosen_pipeline.predict(X_test)
residuals = y_test.values - y_pred

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

lo, hi = min(y_test.min(), y_pred.min()), max(y_test.max(), y_pred.max())
axes[0].plot([lo, hi], [lo, hi], '--', color='#999999', linewidth=1)
axes[0].scatter(y_test, y_pred, s=20, color='#5b8def', alpha=0.7, edgecolors='none')
axes[0].set_xlabel('Actual deduction_pct')
axes[0].set_ylabel('Predicted deduction_pct')
axes[0].set_title('Predicted vs. actual')

axes[1].axhline(0, color='#999999', linewidth=1)
axes[1].scatter(X_test['txn_amount'], residuals, s=20, color='#2bd6a3', alpha=0.7, edgecolors='none')
axes[1].set_xlabel('txn_amount')
axes[1].set_ylabel('Residual (actual - predicted)')
axes[1].set_title('Residuals vs. txn_amount')

axes[2].hist(residuals, bins=15, color='#5b8def', alpha=0.85, edgecolor='white')
axes[2].axvline(0, color='#f04f61', linestyle='--', linewidth=1.2)
axes[2].set_xlabel('Residual')
axes[2].set_title('Residual distribution')

fig.tight_layout()
plt.show()
```

## 9. Refit on the full dataset and save

Mirrors `train.py`. This notebook saves its own copy under `./artifacts/` — it does **not** overwrite the canonical `backend/forecaster/models/` artifact, which only `python -m backend.forecaster.train` produces.

```python
chosen_pipeline.fit(X, y)

out_dir = Path('artifacts')
out_dir.mkdir(exist_ok=True)
out_path = out_dir / 'model_b_deduction_pct.joblib'
joblib.dump(chosen_pipeline, out_path)
print(f'Saved {out_path.resolve()}')
```
